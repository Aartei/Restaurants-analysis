```python
import pandas as pd

df = pd.read_csv("dataset.csv")   # your real file name here
print(df["ratings"].head(10))
```

```python
df["ratings"] = pd.to_numeric(df["ratings"], errors="coerce")
print(df["ratings"].dtype)
print(df["ratings"].mean())
```

```python
print(df["ratings"].isna().sum())
```

```python
rated = df.dropna(subset=["ratings"]).copy()
print(len(df), "->", len(rated))
```

```python
print(rated.groupby("price for one")["ratings"].mean())
```

```python
print(rated.groupby("price for one")["ratings"].agg(["mean", "count"]))
```

```python
import matplotlib.pyplot as plt

rated.groupby("price for one")["ratings"].mean().plot(kind="bar")
plt.title("Average rating by price")
plt.ylabel("Average rating")
plt.show()
```

```python
print(rated.groupby("price for one")["ratings"].agg(["mean", "count"]))
```
